Cleaning Of Data

Step 2 — First cell: Import Pandas

```python
import pandas as pd
```

Step 3 — Load your CSV

```python
df = pd.read_csv("healthcare_dataset.csv")

print("Dataset loaded successfully")
```

Step 4 — Check the dataset

```python
print("Rows and Columns:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
display(df.head())
```

Step 5 — Check missing values

```python
print(df.isnull().sum())
```

Step 6 — Check duplicate records

```python
print("Duplicate rows:", df.duplicated().sum())
```

Step 7 — Remove duplicates

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

Step 8 — Clean column names

```python
df.columns = df.columns.str.strip()

print(df.columns.tolist())
```

Step 9 — Clean text data

```python
text_columns = [
    "Name",
    "Gender",
    "Blood Type",
    "Medical Condition",
    "Doctor",
    "Hospital",
    "Insurance Provider",
    "Admission Type",
    "Medication",
    "Test Results"
]

for column in text_columns:
    df[column] = df[column].astype(str).str.strip()
```

Step 10 — Convert dates

```python
df["Date of Admission"] = pd.to_datetime(
    df["Date of Admission"],
    errors="coerce"
)

df["Discharge Date"] = pd.to_datetime(
    df["Discharge Date"],
    errors="coerce"
)
```

```python
print(df[["Date of Admission", "Discharge Date"]].dtypes)
```

Step 11 — Check invalid dates

```python
print("Invalid admission dates:",
      df["Date of Admission"].isna().sum())

print("Invalid discharge dates:",
      df["Discharge Date"].isna().sum())
```

Step 12 — Check numerical columns

```python
print(df[[
    "Age",
    "Billing Amount",
    "Room Number"
]].describe())
```

Step 13 — Remove invalid ages

```python
df = df[
    (df["Age"] >= 0) &
    (df["Age"] <= 120)
]
```

Step 14 — Remove invalid billing values

```python
df = df[df["Billing Amount"] >= 0]
```

Step 15 — Create Length of Stay

```python
df["Length of Stay"] = (
    df["Discharge Date"] -
    df["Date of Admission"]
).dt.days
```

```python
display(
    df[
        [
            "Date of Admission",
            "Discharge Date",
            "Length of Stay"
        ]
    ].head()
)
```

Step 16 — Remove invalid Length of Stay

```python
df = df[df["Length of Stay"] >= 0]
```

Step 17 — Final missing-value check

```python
print("Missing values after cleaning:")

print(df.isnull().sum())
```

Step 18 — Final duplicate check

```python
print("Duplicate rows after cleaning:",
      df.duplicated().sum())
```

Step 19 — Check final dataset

```python
print("Final Dataset Shape:", df.shape)

display(df.head())
```

Step 20 — Save the cleaned dataset

```python
df.to_csv(
    "cleaned_healthcare_data.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```

Analytics Plot

1. Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

2. Load the cleaned dataset

```python
df = pd.read_csv("cleaned_healthcare_data.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

Plot 1 — Patients by Medical Condition

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y="Medical Condition",
    order=df["Medical Condition"].value_counts().index
)

plt.title("Number of Patients by Medical Condition")
plt.xlabel("Number of Patients")
plt.ylabel("Medical Condition")

plt.tight_layout()
plt.show()
```

Plot 2 — Gender Distribution

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Gender"
)

plt.title("Patient Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")

plt.tight_layout()
plt.show()
```

Plot 3 — Admission Type

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Admission Type",
    order=df["Admission Type"].value_counts().index
)

plt.title("Patients by Admission Type")
plt.xlabel("Admission Type")
plt.ylabel("Number of Patients")

plt.tight_layout()
plt.show()
```

Plot 4 — Patient Test Results

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Test Results",
    order=df["Test Results"].value_counts().index
)

plt.title("Patient Test Results")
plt.xlabel("Test Result")
plt.ylabel("Number of Patients")

plt.tight_layout()
plt.show()
```

Plot 5 — Age Distribution

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Age",
    bins=20,
    kde=True
)

plt.title("Patient Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")

plt.tight_layout()
plt.show()
```

Plot 6 — Billing Amount Distribution

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Billing Amount",
    bins=30,
    kde=True
)

plt.title("Hospital Billing Amount Distribution")
plt.xlabel("Billing Amount")
plt.ylabel("Number of Patients")

plt.tight_layout()
plt.show()
```

Plot 7 — Average Billing by Medical Condition

```python
avg_billing = (
    df.groupby("Medical Condition")["Billing Amount"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=avg_billing.values,
    y=avg_billing.index
)

plt.title("Average Billing Amount by Medical Condition")
plt.xlabel("Average Billing Amount")
plt.ylabel("Medical Condition")

plt.tight_layout()
plt.show()
```

Plot 8 — Average Length of Stay by Condition

```python
avg_stay = (
    df.groupby("Medical Condition")["Length of Stay"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=avg_stay.values,
    y=avg_stay.index
)

plt.title("Average Length of Stay by Medical Condition")
plt.xlabel("Average Length of Stay (Days)")
plt.ylabel("Medical Condition")

plt.tight_layout()
plt.show()
```

Plot 9 — Monthly Patient Admissions

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])

monthly_admissions = (
    df.set_index("Date of Admission")
    .resample("ME")
    .size()
)

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_admissions.index,
    monthly_admissions.values,
    marker="o"
)

plt.title("Monthly Patient Admissions")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Plot 10 — Correlation Heatmap

```python
numeric_columns = [
    "Age",
    "Billing Amount",
    "Room Number",
    "Length of Stay"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Between Patient Variables")

plt.tight_layout()
plt.show()
```

Basic Charts — Matplotlib

A. Line Chart — Monthly Admissions

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("cleaned_healthcare_data.csv")

df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])

monthly = (
    df.set_index("Date of Admission")
    .resample("ME")
    .size()
)

plt.figure(figsize=(10, 5))

plt.plot(
    monthly.index,
    monthly.values,
    marker="o"
)

plt.title("Monthly Patient Admissions")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
